```python
import pickle
import os

spectro_folder = 'spectrograms2'

# Laad de LOG_SPEC_SEGMENTS weer in
try:
    with open(os.path.join(spectro_folder, 'log_spec_segments.pkl'), 'rb') as f:
        # Voeg encoding='latin1' toe
        geladen_segmenten = pickle.load(f, encoding='latin1')
    
    # Nu is 'geladen_segmenten' precies dezelfde lijst als 'LOG_SPEC_SEGMENTS' was
    print(f"Er zijn {len(geladen_segmenten)} segmenten geladen.")

except FileNotFoundError:
    print(f"FOUT: Bestand niet gevonden.")
    print(f"Ik zocht hier: {os.path.abspath(os.path.join(spectro_folder, 'log_spec_segments.pkl'))}")
except pickle.UnpicklingError:
    print(f"FOUT: Het pickle-bestand is corrupt of kan niet gelezen worden.")
except EOFError:
    print(f"FOUT: Het bestand 'log_spec_segments.pkl' is leeg (0 bytes).")
    print("Draai het script dat het bestand opslaat opnieuw.")
except Exception as e:
    print(f"Er is een onverwachte fout opgetreden: {e}")
```
